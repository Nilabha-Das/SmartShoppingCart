```python
master_products = products.merge(nutrition, on="product_id", how="left").merge(attributes, on="product_id", how="left")
master_products.head()
```

```python
master_products.shape
```

```python
master_products.to_csv("../data/phase1/master_products.csv", index = False)
```

```python
user_items = transactions[["customer_id", "product_id", "quantity", "purchase_date"]].copy()
user_items.head()
```

```python
user_items[["customer_id", "product_id"]].drop_duplicates().shape
```

```python
popular_products = (transactions.groupby("product_id")["quantity"].sum().sort_values(ascending = False))
popular_products.head(10)
```

```python
popular_recommendations = (popular_products.head(10).reset_index().merge(master_products, on="product_id", how="left"))
popular_recommendations[["product_id", "product_name", "category", "brand", "price", "rating", "quantity"]]
```

```python
def recommend_popular(n=5):
    return popular_recommendations[["product_id", "product_name", "category", "brand", "price", "rating", "quantity"]].head(n)
```

```python
recommend_popular(5)
```

```python
popular_recommendations.to_csv("../models/popular_baseline.csv", index=False)
```
